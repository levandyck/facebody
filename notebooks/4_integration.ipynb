# DNN integration analysis
- Approach
    - Compute contrasts from condition means per unit type.
    - Face coefficient: Faces - Backgrounds
    - Body coefficient: Bodies - Backgrounds
    - Integration coefficient: Whole persons - Faces - Bodies + Backgrounds
- Face-body integration dataset
    - 100 images of corresponding whole-persons, headless bodies, bodiless heads (Hemera Photo objects)

```python
from myutils.models import ModelLoader
from myutils.dataset import ImageDataLoader
from myutils.feature_extractor import FeatureExtractor
from facebody.config import DATA_ROOT
from facebody.integration import integration_analysis, test_coefs_within
from facebody.plotting import plot_integration_coefs, plot_integration_coef_layers

device = "cuda:0"
```

```python
# Specify model and layers
model_name = "alexnet_ecoset"
layers = ["conv1", "conv2", "conv3", "conv4", "conv5", "fc6", "fc7"]
```

## Face-body integration analysis

```python
# Run integration analysis
il = ImageDataLoader(root=DATA_ROOT / "datasets" / "integration")
ml = ModelLoader(model_name, DATA_ROOT, device)
fe = FeatureExtractor(ml, layers)
activs, labels = fe.extract(il)
integration_analysis(model_name, layers, activs, labels, class_to_idx=il.dataset.class_to_idx)
```

```python
# Fig. 4: Plot coefficients per unit type in penultimate layer
layer = "fc7"
plot_integration_coefs(model_name, layer, out_path="main/integration_coefs.pdf")

# Stats: Bootstrapping
for coef in ["beta_F", "beta_B", "beta_FB"]:
    test_coefs_within(model_name, layer, coef)
```

```python
# Fig. 4C: Plot coefficients per unit type across layers
plot_integration_coef_layers(
    model_name, layers,
    out_path="main/integration_coef_layers.pdf"
)
```

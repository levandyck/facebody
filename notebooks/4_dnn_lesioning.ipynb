# DNNs - Lesioning analysis
Train linear classifier for different person perception tasks, lesion each unit type, and measure performance drop

```python
from facebody.model_sets import ENCODING_MODEL, LESIONING_LAYER, LESIONING_TASKS
from facebody.dnn.lesioning import run_lesioning_analysis, report_lesioning
from facebody.plotting import plot_lesioning_drop

device = "cuda:0"
```

```python
# Model, readout layer, and tasks
MODEL = ENCODING_MODEL
LAYER = LESIONING_LAYER
TASKS = LESIONING_TASKS
```

## Lesioning analysis
- Person recognition datasets
    - Faces: VGGFace2 test set (Cao et al., 2017)
    - Persons: Celeb-reID training set (Huang et al., 2019)
    - Actions: Stanford 40 Actions (Yao et al., 2011)

```python
# Run lesioning analysis per task
for task in TASKS:
    run_lesioning_analysis(MODEL, LAYER, task, device)
```

```python
# Fig. 5: Plot normalized accuracy drop per unit type
plot_lesioning_drop(MODEL, TASKS, out_path="main/lesioning.pdf")
```

```python
# Table S1: Lesioning results
for task in TASKS:
    for use in ["abs", "rel"]:
        report_lesioning(MODEL, task, use=use)
```

# DNN lesioning analysis
- Approach
    - Train linear classifier for different tasks on frozen DNN backbone
    - Measure task performance before and after lesioning different unit types = norm. accuracy drop
- Person recognition datasets
    - Faces: VGGFace2 test set (Cao et al., 2017)
    - Persons: Celeb-reID training set (Huang et al., 2019)
    - Actions: Stanford 40 Actions (Yao et al., 2011)

```python
from facebody.lesioning import run_lesioning_analysis, print_lesioning_summary
from facebody.plotting import plot_lesioning_drop

device = "cuda:0"
```

```python
# Specify model, penultimate layer, and tasks
model_name = "alexnet_ecoset"
layer = "relu7"
tasks = ["face", "person", "action"]
```

## Lesioning analysis

```python
# Run lesioning analysis per task
for task in tasks:
    run_lesioning_analysis(model_name, readout_layer=layer, device=device, task=task)
```

```python
# Fig. 3: Plot normalized accuracy drop per unit type
plot_lesioning_drop(model_name, tasks, out_path="main/lesioning.pdf")
```

```python
# Stats: Lesioning summary and bootstrap tests
for task in tasks:
    print_lesioning_summary(model_name, task)
```

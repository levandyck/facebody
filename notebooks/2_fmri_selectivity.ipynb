# fMRI - Selective voxels
Identify and validate selective voxels in visual cortex

```python
from facebody.config import PROJECT_ROOT, SUBJECTS
from facebody.fmri.floc import run_floc_glm, run_fmri_selectivity
from facebody.fmri.gradient import build_gradient, run_fmri_profile
from facebody.plotting.flatmaps import plot_roi_flatmaps_subjects
from facebody.plotting import plot_perc_ventral, stats_perc_ventral
from facebody.myutils.nsd import SubjectLoader
```

```python
# NSD participants
print(SUBJECTS)
```

### Functional localizer

```python
# Run fMRI GLM
run_floc_glm(SUBJECTS)
```

```python
# Find selective voxels
run_fmri_selectivity(SUBJECTS)
```

```python
# Fig. S3: Plot flatmaps with selective voxels
plot_roi_flatmaps_subjects(SUBJECTS, out_path="suppl/fmri_sel_flatmaps_subjects.pdf")
```

### Ventral visual pathway

```python
# Build posterior -> anterior gradient, one axis per hemisphere
for subj in SUBJECTS:
    sl = SubjectLoader(subj, PROJECT_ROOT)
    build_gradient(subj, sl, PROJECT_ROOT)
```

```python
# Profile selectivity along gradient, pooled across hemispheres
run_fmri_profile(SUBJECTS, PROJECT_ROOT)
```

```python
# Fig. 1D: Plot percentage of selective voxels across ventral pathway
plot_perc_ventral(SUBJECTS, out_path="main/perc_ventral.pdf")
stats_perc_ventral()
```
